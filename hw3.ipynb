## HW คำนวณเวลาเป็นวินาทีของเวลาต่อไปนี้โดยใช้คำสั่ง split() ช่วย (print ออกมาให้สวยงาม)
12:30:15

13:41:07

12:53:15

00:59:25

11:11:11

16:06:09

````python
time = '12:30:15'
t = time.split(':')
h, m, se = int(t[0]), int(t[1]), int(t[2])
total_se = (h * 3600) + (m * 60) + se
print(f"เวลา {time} เท่ากับ {total_se:,} วินาที\n")

time = '13:41:07'
t = time.split(':')
h, m, se = int(t[0]), int(t[1]), int(t[2])
total_se = (h * 3600) + (m * 60) + se
print(f"เวลา {time} เท่ากับ {total_se:,} วินาที\n")

time = '12:53:15'
t = time.split(':')
h, m, se = int(t[0]), int(t[1]), int(t[2])
total_se = (h * 3600) + (m * 60) + se
print(f"เวลา {time} เท่ากับ {total_se:,} วินาที\n")

time = '00:59:25'
t = time.split(':')
h, m, se = int(t[0]), int(t[1]), int(t[2])
total_se = (h * 3600) + (m * 60) + se
print(f"เวลา {time} เท่ากับ {total_se:,} วินาที\n")

time = '11:11:11'
t = time.split(':')
h, m, se = int(t[0]), int(t[1]), int(t[2])
total_se = (h * 3600) + (m * 60) + se
print(f"เวลา {time} เท่ากับ {total_se:,} วินาที\n")

time = '16:06:09'
t = time.split(':')
h, m, se = int(t[0]), int(t[1]), int(t[2])
total_se = (h * 3600) + (m * 60) + se
print(f"เวลา {time} เท่ากับ {total_se:,} วินาที\n")
````

## ให้สร้างฟังก์ชั่นคำนวณเวลาเป็นวินาทีของเวลาใดๆ และ print ออกมาให้สวยงาม

````python
def second(time):
    t = time.split(':')
    h = int(t[0])
    m = int(t[1])
    se = int(t[2])
    total_second = (h * 3600) + (m * 60) + se
    print(f"เวลา {time} ชั่วโมง เท่ากับ {total_second:,} วินาที")
````

````python
second('12:30:15')
second('13:41:07')
second('12:53:15')
second('00:59:25')
second('11:11:11')
````

## ให้ สร้าง list ของ เวลา ```['12:30:15','13:41:07',....]``` แล้ววนลูปเรียกฟังก์ชั่นคำนวณเวลาเป็น วินาที

**commit ว่า HW3**

````python
time_list = [
    "12:30:15",
    "13:41:07",
    "12:53:15",
    "00:59:25",
    "11:11:11",
    "16:06:09"]
````

````python
print(f'< คำนวณเวลาจาก ชั่วโมง เป็น วินาที >')
for i in time_list:
    second(i)
````
